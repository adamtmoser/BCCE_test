# PDB Visualization with py3Dmol

```python
# Install py3Dmol library for 3D molecular visualization
!pip install -q py3Dmol
```

```python
# Import py3Dmol
import py3Dmol

# Define the PDB file to be visualized
pdb_filename = '1CGO.pdb'
```

```python
# Read the PDB file content
with open(pdb_filename, 'r') as f:
    pdb_content = f.read()

# Create a py3Dmol viewer with specified dimensions
view = py3Dmol.view(width=800, height=600)

# Add the PDB model to the viewer
view.addModel(pdb_content, 'pdb')

# Apply a cartoon style visualization with a spectrum color scheme
view.setStyle({'cartoon': {'color': 'spectrum'}})

# Zoom to fit the entire model within the view
view.zoomTo()

# Display the py3Dmol visualization
view.show()
```
